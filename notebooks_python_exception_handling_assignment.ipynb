## Create a class-based bank account system. Handle deposits and withdrawals using custom exceptions for invalid transaction amounts and insufficient balance.

```c++
class InvalidAmountError(Exception): pass
class InsufficientBalanceError(Exception): pass

class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance
    def deposit(self, amount):
        if amount <= 0:
            raise InvalidAmountError("Deposit must be positive.")
        self.balance += amount
    def withdraw(self, amount):
        if amount <= 0:
            raise InvalidAmountError("Withdrawal must be positive.")
        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient balance.")
        self.balance -= amount

acc = BankAccount(1000)
try:
    acc.withdraw(200)
    print(f"Balance: {acc.balance}")
except (InvalidAmountError, InsufficientBalanceError) as e:
    print(e)
```

## Create an e-commerce order system that handles invalid products, invalid quantities, insufficient stock, and invalid payments. Use custom exceptions for business-rule failures.

```c++
class InvalidProductError(Exception): pass
class InvalidQuantityError(Exception): pass
class InsufficientStockError(Exception): pass
class InvalidPaymentError(Exception): pass

stock = {"Laptop": 5, "Mobile": 10}
prices = {"Laptop": 50000, "Mobile": 20000}
try:
    product = input("Product: ")
    qty = int(input("Quantity: "))
    payment = float(input("Payment: "))
    if product not in stock:
        raise InvalidProductError("Product not found.")
    if qty <= 0:
        raise InvalidQuantityError("Quantity must be positive.")
    if qty > stock[product]:
        raise InsufficientStockError("Not enough stock.")
    total = prices[product]*qty
    if payment < total:
        raise InvalidPaymentError(f"Total is {total}")
    print("Order placed!")
except (InvalidProductError, InvalidQuantityError, InsufficientStockError, InvalidPaymentError, ValueError) as e:
    print(f"Failed: {e}")
```

## ### University Course Registration System

### Goal:
Create a program that allows students to register for university courses.

### Steps:
- Store students and courses.
- Check whether the student exists.
- Check whether the course exists.
- Prevent duplicate registration.
- Check course capacity.
- Check prerequisites.
- Create and handle suitable custom exceptions.

### Example:
Student ID: 102
Course: Python
→ Registration successful

```c++
class StudentNotFoundError(Exception): pass
class CourseNotFoundError(Exception): pass
class DuplicateRegistrationError(Exception): pass
class CourseFullError(Exception): pass
class PrerequisiteError(Exception): pass

students = {"102": {"name": "Ali", "courses": []}, "103": {"name": "Sara", "courses": ["Python"]}}
courses = {"Python": {"capacity": 2, "enrolled": 1, "prereq": None}, "Advanced Python": {"capacity": 2, "enrolled": 0, "prereq": "Python"}}

def register(sid, course):
    if sid not in students:
        raise StudentNotFoundError("Student not found!")
    if course not in courses:
        raise CourseNotFoundError("Course not found!")
    if course in students[sid]["courses"]:
        raise DuplicateRegistrationError("Already registered!")
    if courses[course]["enrolled"] >= courses[course]["capacity"]:
        raise CourseFullError("Course full!")
    prereq = courses[course]["prereq"]
    if prereq and prereq not in students[sid]["courses"]:
        raise PrerequisiteError(f"Prerequisite '{prereq}' required!")
    students[sid]["courses"].append(course)
    courses[course]["enrolled"] += 1
    return "Registration successful!"

try:
    print(register(input("Student ID: "), input("Course: ")))
except (StudentNotFoundError, CourseNotFoundError, DuplicateRegistrationError, CourseFullError, PrerequisiteError) as e:
    print(f"Failed: {e}")
```

## ### Hospital Appointment System

### Goal:
Create a program that manages hospital appointments.

### Steps:
- Store patients and doctors.
- Validate patient and doctor information.
- Check doctor availability.
- Prevent duplicate appointment times.
- Create custom exceptions for important business errors.
- Handle invalid input without crashing.

### Example:
Patient ID: P102
Doctor: D05
Time: 10:30 AM
→ Appointment booked successfully

```c++
class PatientNotFoundError(Exception): pass
class DoctorNotFoundError(Exception): pass
class DoctorNotAvailableError(Exception): pass
class DuplicateAppointmentError(Exception): pass

patients = {"P101": "Ali", "P102": "Sara"}
doctors = {"D05": "Dr. Ahmed", "D06": "Dr. Ayesha"}
appointments = {"D05": ["09:00 AM"], "D06": []}

def book(pid, did, time):
    if pid not in patients:
        raise PatientNotFoundError("Patient not found!")
    if did not in doctors:
        raise DoctorNotFoundError("Doctor not found!")
    if time in appointments.get(did, []):
        raise DuplicateAppointmentError("Already booked at this time!")
    if time not in ["09:00 AM", "10:30 AM", "11:00 AM", "02:00 PM"]:
        raise DoctorNotAvailableError("Doctor not available!")
    appointments[did].append(time)
    return "Appointment booked successfully!"

try:
    print(book(input("Patient ID: "), input("Doctor ID: "), input("Time: ")))
except (PatientNotFoundError, DoctorNotFoundError, DoctorNotAvailableError, DuplicateAppointmentError) as e:
    print(f"Failed: {e}")
```

## ### Digital Wallet System

### Goal:
Create a class-based digital wallet application.

### Steps:
- Validate wallet ID and PIN.
- Handle inactive accounts.
- Validate transfer amounts.
- Check balance.
- Enforce a transfer limit.
- Create multiple custom exceptions.
- Make sure failed transfers do not change wallet balances.

### Example:
Transfer: Rs. 2,000
→ Transfer successful

```c++
class InvalidWalletError(Exception): pass
class InvalidPinError(Exception): pass
class InactiveAccountError(Exception): pass
class InvalidAmountError(Exception): pass
class InsufficientBalanceError(Exception): pass
class TransferLimitError(Exception): pass

class Wallet:
    def __init__(self, wid, pin, balance, active=True):
        self.wid, self.pin, self.balance, self.active = wid, pin, balance, active
    def transfer(self, to_wallet, amount, pin, limit=10000):
        if not self.active or not to_wallet.active:
            raise InactiveAccountError("Wallet inactive!")
        if pin != self.pin:
            raise InvalidPinError("Wrong PIN!")
        if amount <= 0:
            raise InvalidAmountError("Amount must be > 0!")
        if amount > limit:
            raise TransferLimitError(f"Limit is Rs. {limit}")
        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient balance!")
        self.balance -= amount
        to_wallet.balance += amount
        return f"Transfer: Rs. {amount} -> Transfer successful!"

wallets = {"W101": Wallet("W101", "1234", 5000), "W102": Wallet("W102", "5678", 2000)}
try:
    from_id = input("From Wallet: ")
    to_id = input("To Wallet: ")
    if from_id not in wallets or to_id not in wallets:
        raise InvalidWalletError("Wallet not found!")
    amount = float(input("Amount: "))
    print(wallets[from_id].transfer(wallets[to_id], amount, input("PIN: ")))
    print(f"New Balance: {wallets[from_id].balance}")
except ValueError:
    print("Invalid amount!")
except (InvalidWalletError, InvalidPinError, InactiveAccountError, InvalidAmountError, InsufficientBalanceError, TransferLimitError) as e:
    print(f"Failed: {e}")
finally:
    print("Wallet transaction ended.")
```

## ### Retail Inventory and Sales System

### Task:
Create a menu-driven inventory and sales application.

The program should allow the user to:
- Add stock
- View inventory
- Sell products
- Search for products

Handle invalid input, missing products, invalid quantities, and insufficient stock. Use `raise`, custom exceptions, `else`, and `finally` where appropriate.

The application should continue running after recoverable errors.

```c++
class ProductNotFoundError(Exception): pass
class InvalidQuantityError(Exception): pass
class InsufficientStockError(Exception): pass

inventory = {"Milk": {"price": 120, "stock": 20}, "Bread": {"price": 80, "stock": 15}}

def add_stock():
    try:
        name = input("Product: ")
        qty = int(input("Qty: "))
        price = float(input("Price: "))
        if qty <= 0 or price <= 0:
            raise InvalidQuantityError("Qty/Price must be positive.")
        inventory[name] = {"price": price, "stock": inventory.get(name, {"stock": 0})["stock"] + qty}
    except ValueError:
        print("Invalid input!")
    except InvalidQuantityError as e:
        print(e)
    else:
        print("Stock added!")
    finally:
        print("Add stock completed.")

def sell():
    try:
        name = input("Product: ")
        qty = int(input("Qty: "))
        if name not in inventory:
            raise ProductNotFoundError("Product not found!")
        if qty <= 0:
            raise InvalidQuantityError("Qty must be positive!")
        if qty > inventory[name]["stock"]:
            raise InsufficientStockError("Not enough stock!")
    except ValueError:
        print("Invalid qty!")
    except (ProductNotFoundError, InvalidQuantityError, InsufficientStockError) as e:
        print(f"Sale failed: {e}")
    else:
        inventory[name]["stock"] -= qty
        print(f"Sold! Total: {inventory[name]['price']*qty}")
    finally:
        print("Sale completed.")

while True:
    print("\n1.Add Stock 2.View 3.Sell 4.Search 5.Exit")
    try:
        ch = int(input("Choice: "))
    except ValueError:
        print("Enter valid number!")
        continue
    if ch == 1:
        add_stock()
    elif ch == 2:
        for k, v in inventory.items():
            print(f"{k} - {v}")
    elif ch == 3:
        sell()
    elif ch == 4:
        name = input("Search: ")
        print(inventory.get(name, "Not found!"))
    elif ch == 5:
        break
    else:
        print("Invalid choice!")
```

## ### Food Delivery Application

### Goal:
Create a complete console-based food delivery application using exception handling.

### Steps:
- Register customers.
- Select a restaurant.
- Display menu items.
- Take item quantities.
- Validate stock.
- Validate minimum order amount.
- Validate delivery area.
- Process payment.
- Create custom exception classes for business-rule failures.
- Use `try`, specific `except` blocks, `else`, `finally`, and `raise`.
- Prevent invalid operations from partially changing important application data.

### Example:
Customer: Ali
Restaurant: ABC Foods
Item: Burger
Quantity: 2
→ Order placed successfully

```c++
class CustomerNotFoundError(Exception): pass
class RestaurantNotFoundError(Exception): pass
class ItemNotFoundError(Exception): pass
class InsufficientStockError(Exception): pass
class MinimumOrderError(Exception): pass
class DeliveryAreaError(Exception): pass
class PaymentFailedError(Exception): pass
class InvalidQuantityError(Exception): pass

customers = {}
restaurants = {
    "ABC Foods": {"menu": {"Burger": {"price": 350, "stock": 10}}, "areas": ["Gulshan", "Clifton"], "min_order": 500},
    "XYZ Biryani": {"menu": {"Biryani": {"price": 400, "stock": 20}}, "areas": ["Gulshan"], "min_order": 300}
}

def register():
    try:
        cid, name, area = input("ID: "), input("Name: "), input("Area: ")
        if not cid or not name or not area:
            raise ValueError("All fields required!")
        customers[cid] = {"name": name, "area": area}
    except ValueError as e:
        print(e)
    else:
        print("Registered!")

def order():
    try:
        cid = input("Customer ID: ")
        if cid not in customers:
            raise CustomerNotFoundError("Customer not found!")
        rname = input(f"Restaurant {list(restaurants.keys())}: ")
        if rname not in restaurants:
            raise RestaurantNotFoundError("Restaurant not found!")
        rest = restaurants[rname]
        print(f"Menu: {rest['menu']}")
        item = input("Item: ")
        if item not in rest["menu"]:
            raise ItemNotFoundError("Item not available!")
        qty = int(input("Quantity: "))
        if qty <= 0:
            raise InvalidQuantityError("Qty must be positive!")
        if qty > rest["menu"][item]["stock"]:
            raise InsufficientStockError("Not enough stock!")
        total = rest["menu"][item]["price"]*qty
        if total < rest["min_order"]:
            raise MinimumOrderError(f"Min order Rs. {rest['min_order']}")
        if customers[cid]["area"] not in rest["areas"]:
            raise DeliveryAreaError("Delivery not available in your area!")
        payment = float(input(f"Payment (Total {total}): "))
        if payment < total:
            raise PaymentFailedError("Payment insufficient!")
        rest["menu"][item]["stock"] -= qty
    except ValueError:
        print("Invalid numeric input!")
    except (CustomerNotFoundError, RestaurantNotFoundError, ItemNotFoundError, InsufficientStockError, MinimumOrderError, DeliveryAreaError, PaymentFailedError, InvalidQuantityError) as e:
        print(f"Order failed: {e}")
    else:
        print(f"Order placed successfully! Customer: {customers[cid]['name']} Restaurant: {rname} Item: {item} Quantity: {qty}")
    finally:
        print("Order process ended.")

while True:
    print("\n1.Register 2.Order 3.Exit")
    try:
        ch = int(input("Choice: "))
    except ValueError:
        print("Enter valid number!")
        continue
    if ch == 1:
        register()
    elif ch == 2:
        order()
    elif ch == 3:
        break
```
